## 概念

影像也是數值資料。平均數描述中心，標準差描述分散；相同平均數不代表相同資料。樣本變異數使用 n−1 作分母，與描述整份有限資料的 n 分母有所不同。

## 先想一想

比較以下兩組假想像素資料：平均數相同時，哪組差異更大？

```python
import numpy as np
import matplotlib.pyplot as plt
a = np.array([100, 100, 100, 100, 100])
b = np.array([0, 50, 100, 150, 200])
for name, values in [('A', a), ('B', b)]:
    print(name, 'mean=', values.mean(), 'median=', np.median(values),
          'sample std=', values.std(ddof=1),
          'quartiles=', np.quantile(values, [0.25, 0.5, 0.75]))
plt.hist([a, b], bins=np.arange(0, 226, 25), label=['A', 'B'])
plt.xlabel('Pixel value'); plt.ylabel('Count'); plt.legend(); plt.show()
assert a.mean() == b.mean()
assert a.std(ddof=1) < b.std(ddof=1)
```

## 如何解讀

兩組平均數都是 100，但 B 的分散程度較高。直方圖比只看平均數提供更多訊息。像素分布仍不包含空間排列資訊，因此無法單憑直方圖辨識影像。
